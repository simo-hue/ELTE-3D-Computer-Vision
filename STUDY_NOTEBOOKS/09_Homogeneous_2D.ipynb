## Idea

Add a third coordinate: $(x,y)\rightarrow(x,y,1)$, and treat $(x,y,w)$ and $(\lambda x,\lambda y,\lambda w)$ as the *same* point. Then translation becomes a matrix product too, and every 2D transformation is a $3\times3$ matrix that you chain by multiplying. The same trick gives the perspective effect: divide by $w$ at the end. Lines are also 3-vectors, and the cross product joins and intersects them.

## Key formulas

## Homogeneous Coordinates (2D)

A 2D point $(x,y)$ can be represented using homogeneous coordinates:

$$
\mathbf{p}=
\begin{bmatrix}
x\\y\\w
\end{bmatrix}
$$

To convert back to Cartesian coordinates, divide by $w$:

$$
(x,y,w)\sim\left(\frac{x}{w},\frac{y}{w}\right),
\qquad w\neq0
$$

If $w=0$, the point represents a **point at infinity**, corresponding to a direction.

## 2D Transformation Matrices

### Translation

Moves a point by $t_x$ and $t_y$:

$$
T=
\begin{bmatrix}
1&0&t_x\\
0&1&t_y\\
0&0&1
\end{bmatrix}
$$

### Scaling

Changes the size along the X and Y axes:

$$
S=
\begin{bmatrix}
s_x&0&0\\
0&s_y&0\\
0&0&1
\end{bmatrix}
$$

### Rotation

Rotates a point around the origin by angle $\theta$:

$$
R=
\begin{bmatrix}
\cos\theta&-\sin\theta&0\\
\sin\theta&\cos\theta&0\\
0&0&1
\end{bmatrix}
$$

### Shearing

Changes the shape by shifting coordinates proportionally:

$$
H_{sh}=
\begin{bmatrix}
1&k&0\\
0&1&0\\
0&0&1
\end{bmatrix}
$$

Here, $k$ controls the horizontal shear.

## Lines in Homogeneous Coordinates

A 2D line:

$$
ax+by+c=0
$$

Can be represented by a vector:

$$
\mathbf{l}=
\begin{bmatrix}
a\\b\\c
\end{bmatrix}
$$

A point $\mathbf{p}$ lies on the line if:

$$
\boxed{\mathbf{l}^T\mathbf{p}=0}
$$

### Constructing a Line from Two Points

Given two points $\mathbf{p}_1$ and $\mathbf{p}_2$ in homogeneous coordinates:

$$
\boxed{
\mathbf{l}=\mathbf{p}_1\times\mathbf{p}_2
}
$$

### Finding the Intersection of Two Lines

Given two lines $\mathbf{l}_1$ and $\mathbf{l}_2$:

$$
\boxed{
\mathbf{p}=\mathbf{l}_1\times\mathbf{l}_2
}
$$

The resulting point is in homogeneous coordinates. If its third coordinate is zero, the lines intersect at infinity (they are parallel in the Euclidean plane).

## Transformation Hierarchy

From most restrictive to most general:

1. **Euclidean:** Rotation and translation.

   $$R,\ t$$

   Preserves distances and angles.

2. **Similarity:** Uniform scaling, rotation and translation.

   $$sR,\ t$$

   Preserves angles and relative proportions.

3. **Affine:** General linear transformation and translation.

   $$A,\ t$$

   Preserves parallelism and ratios along the same line.

   Homogeneous matrix has last row:

   $$[0,\ 0,\ 1]$$

4. **Projective:** General invertible $3\times3$ matrix.

   $$H$$

   Preserves straight lines and incidence, but not necessarily parallelism or angles.

The hierarchy is:

$$
\boxed{
\text{Euclidean}
\subset
\text{Similarity}
\subset
\text{Affine}
\subset
\text{Projective}
}
$$

# Worked example 1 (on paper): rotate about a point

Rotate $\mathbf{p}=(2,0)$ by $90^\circ$ about $\mathbf{c}=(1,1)$: move $\mathbf{c}$ to the origin, rotate, move back. Read the product from right to left: $$M=T(1,1)\,R(90^\circ)\,T(-1,-1),\quad
R\,T(-1,-1)=\begin{bmatrix}0&-1&0\\1&0&0\\0&0&1\end{bmatrix}\begin{bmatrix}1&0&-1\\0&1&-1\\0&0&1\end{bmatrix}=\begin{bmatrix}0&-1&1\\1&0&-1\\0&0&1\end{bmatrix}$$ $$M=\begin{bmatrix}1&0&1\\0&1&1\\0&0&1\end{bmatrix}\begin{bmatrix}0&-1&1\\1&0&-1\\0&0&1\end{bmatrix}=\begin{bmatrix}0&-1&2\\1&0&0\\0&0&1\end{bmatrix},\qquad
M\begin{bmatrix}2\\0\\1\end{bmatrix}=\begin{bmatrix}2\\2\\1\end{bmatrix}\Rightarrow(2,2)$$ Check: $\mathbf{p}-\mathbf{c}=(1,-1)$, and rotating it by $90^\circ$ gives $(1,1)$. Adding $\mathbf{c}$ back gives $(2,2)$ ✓.

# Worked example 2 (on paper): lines with the cross product

Line through $(0,0,1)$ and $(1,1,1)$: $\ \mathbf{l}_1=(0\cdot1-1\cdot1,\ 1\cdot1-0\cdot1,\ 0\cdot1-0\cdot1)=(-1,1,0)$, i.e. $-x+y=0$.\
Line $x=2$: $\mathbf{l}_2=(1,0,-2)$. Intersection: $\ \mathbf{l}_1\times\mathbf{l}_2=(1\cdot(-2)-0\cdot0,\ 0\cdot1-(-1)(-2),\ (-1)\cdot0-1\cdot1)=(-2,-2,-1)\sim(2,2)$.\
The parallel lines $x=0$ and $x=2$ intersect at $(0,2,0)$: a point at infinity in the $y$ direction.

# The same in Python / NumPy

```python
import numpy as np

def T(tx, ty):                       # translation
    return np.array([[1, 0, tx], [0, 1, ty], [0, 0, 1]], dtype=float)

def R(deg):                          # rotation about the origin
    a = np.radians(deg)
    return np.array([[np.cos(a), -np.sin(a), 0], [np.sin(a), np.cos(a), 0], [0, 0, 1]])

# rotate by 90 deg about the centre (1, 1): move centre to origin, rotate, move back
M = T(1, 1) @ R(90) @ T(-1, -1)
print("M =\n", np.round(M, 10))
p = np.array([2, 0, 1.0])            # homogeneous point
q = M @ p
print("M @ (2,0,1) =", np.round(q, 10), "-> (x, y) =", np.round(q[:2] / q[2], 10))

# lines and points with the cross product
a, b = np.array([0, 0, 1.0]), np.array([1, 1, 1.0])
l1 = np.cross(a, b)                  # line through two points
l2 = np.array([1, 0, -2.0])          # line x = 2   (1*x + 0*y - 2 = 0)
x = np.cross(l1, l2)                 # intersection of two lines
print("line through (0,0),(1,1):", l1)
print("intersection with x=2   :", x, "->", x[:2] / x[2])

# parallel lines meet at a point at infinity (last coordinate 0)
print("x=0 and x=2 meet at     :", np.cross([1, 0, 0.0], [1, 0, -2.0]))

# perspective (projective) matrix: division by w is required
H = np.array([[1, 0, 0], [0, 1, 0], [0.5, 0, 1.0]])
for pt in [(0, 0), (1, 0), (2, 0)]:
    v = H @ np.array([*pt, 1.0])
    print(pt, "->", np.round(v[:2] / v[2], 4), "  (w =", v[2], ")")
```
