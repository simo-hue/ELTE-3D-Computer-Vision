**00  Study plan — 3D Computer Vision**\
ELTE, 2026 fall (Hajder, Ahadli, Tófalvi)

# 1. Course facts (from Lecture 00)

-   **Grade** = 50% oral exam + 50% assignments. You need at least 20% from *each* part.\
    Final grade: 5 from 85%, 4 from 70%, 3 from 55%, 2 from 40%.

-   **Practice:** two or three assignments during the term, plus extra points from the lab teacher.

-   **Quizzes** at the start of the labs: 10 in total, 5 questions each. 4–5 correct gives $+1$, 3 correct gives $0$, fewer gives $-1$.\
    One retake is possible in the first week of the exam period (25 questions, 17 correct needed).

-   **Lecture topics:**

    -   estimation: homogeneous and inhomogeneous systems, least squares, RANSAC;

    -   point clouds: line, plane, circle and sphere fitting;

    -   transformations and homography; camera calibration;

    -   stereo: F, E, triangulation, planar motion;

    -   multi-view: bundle adjustment, Tomasi–Kanade.
