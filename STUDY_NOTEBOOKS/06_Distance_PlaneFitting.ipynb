## Idea

A line ($ax+by+c=0$) or a plane ($ax+by+cz+d=0$) is written as $\mathbf{l}^T\mathbf{x}+m=0$. The **distance** of a point to it follows from a small Lagrange problem. The **best** line or plane through many noisy points (smallest sum of squared *perpendicular* distances) has two parts: it passes through the **centroid**, and its normal is the **smallest eigenvector** of the scatter matrix.

## Key formulas

## Point-to-Plane Distance

For a plane defined by:

$$
\mathbf{l}^T\mathbf{x}+m=0
$$

The perpendicular distance from a point $\mathbf{x}_0$ to the plane is:

$$
\boxed{
d(\mathbf{x}_0)=
\frac{|\mathbf{l}^T\mathbf{x}_0+m|}
{\sqrt{\mathbf{l}^T\mathbf{l}}}
}
$$

If the normal vector is already normalized ($\|\mathbf{l}\|=1$):

$$
d(\mathbf{x}_0)=
|\mathbf{l}^T\mathbf{x}_0+m|
$$

## Best-Fit Plane Using Least Squares (PCA)

Given $N$ points, find the plane that minimizes the sum of squared perpendicular distances:

$$
\min_{\mathbf{n},d}
\sum_{i=1}^{N}
(\mathbf{n}^T\mathbf{x}_i+d)^2
$$

Subject to the unit-normal constraint:

$$
\|\mathbf{n}\|=1
$$

**Step 1: Calculate the centroid**

$$
\bar{\mathbf{x}}=
\frac{1}{N}\sum_{i=1}^{N}\mathbf{x}_i
$$

**Step 2: Center the points**

Construct matrix $X$ by subtracting the centroid from every point:

$$
X=
\begin{bmatrix}
(\mathbf{x}_1-\bar{\mathbf{x}})^T\\
(\mathbf{x}_2-\bar{\mathbf{x}})^T\\
\vdots\\
(\mathbf{x}_N-\bar{\mathbf{x}})^T
\end{bmatrix}
$$

**Step 3: Find the normal vector**

The normal is the eigenvector corresponding to the smallest eigenvalue of $X^TX$:

$$
\boxed{
\mathbf{n}=
\text{smallest eigenvector of }X^TX
}
$$

**Step 4: Calculate the plane constant**

$$
\boxed{
d=-\mathbf{n}^T\bar{\mathbf{x}}
}
$$

**Bonus fact:**

The minimum sum of squared perpendicular distances equals the smallest eigenvalue of $X^TX$:

$$
\boxed{
\text{Minimum cost}=
\lambda_{\min}(X^TX)
}
$$

## Where Does the Point-to-Plane Distance Formula Come From?

We want to find the closest point $\mathbf{x}$ on the plane to a given point $\mathbf{x}_0$.

Minimize the squared Euclidean distance:

$$
(\mathbf{x}-\mathbf{x}_0)^T
(\mathbf{x}-\mathbf{x}_0)
$$

Subject to the plane constraint:

$$
\mathbf{l}^T\mathbf{x}+m=0
$$

**Step 1: Define the Lagrangian**

$$
J=
(\mathbf{x}-\mathbf{x}_0)^T
(\mathbf{x}-\mathbf{x}_0)
+\lambda(\mathbf{l}^T\mathbf{x}+m)
$$

**Step 2: Differentiate with respect to $\mathbf{x}$**

$$
2(\mathbf{x}-\mathbf{x}_0)+\lambda\mathbf{l}
=\mathbf{0}
$$

Rearrange:

$$
\mathbf{x}=
\mathbf{x}_0-\frac{\lambda}{2}\mathbf{l}
$$

**Step 3: Apply the plane constraint**

$$
\mathbf{l}^T
\left(\mathbf{x}_0-\frac{\lambda}{2}\mathbf{l}\right)+m=0
$$

Solving for $\lambda$:

$$
\lambda=
\frac{2(\mathbf{l}^T\mathbf{x}_0+m)}
{\mathbf{l}^T\mathbf{l}}
$$

**Step 4: Calculate the distance**

The distance is the length of $\mathbf{x}-\mathbf{x}_0$:

$$
\begin{aligned}
d(\mathbf{x}_0)
&=\|\mathbf{x}-\mathbf{x}_0\|\\
&=\left\|-\frac{\lambda}{2}\mathbf{l}\right\|\\
&=\frac{|\mathbf{l}^T\mathbf{x}_0+m|}
{\sqrt{\mathbf{l}^T\mathbf{l}}}
\end{aligned}
$$

## Why Do We Use the Centroid?

When minimizing the sum of squared distances, differentiating with respect to the plane constant $d$ gives:

$$
\frac{\partial}{\partial d}
\sum_{i=1}^{N}
(\mathbf{n}^T\mathbf{x}_i+d)^2=0
$$

Solving:

$$
\boxed{
d=-\mathbf{n}^T\bar{\mathbf{x}}
}
$$

This means the best-fit plane passes through the centroid of the points.

# Worked example 2 (on paper): best line through 4 points

Points $(1,2),(2,3),(3,5),(4,6)$.

**1. Centroid.** $\bar{\mathbf{x}}=\left(\tfrac{10}4,\tfrac{16}4\right)=(2.5,\,4)$.

**2. Centre the points.** $(-1.5,-2),\ (-0.5,-1),\ (0.5,1),\ (1.5,2)$.

**3. Scatter matrix.** $\sum x^2=2.25+0.25+0.25+2.25=5$, $\sum xy=3+0.5+0.5+3=7$, $\sum y^2=4+1+1+4=10$:   $S=\begin{bmatrix}5&7\\7&10\end{bmatrix}$.

**4. Eigenvalues.** $\lambda^2-15\lambda+(50-49)=0\Rightarrow\lambda=\frac{15\pm\sqrt{221}}2$, so $\lambda_{\min}\approx0.0670$ and $\lambda_{\max}\approx14.933$.

**5. Normal.** $(5-0.067)n_1+7n_2=0\Rightarrow\mathbf{n}\propto(7,\,-4.933)$. Its norm is $\sqrt{49+24.33}\approx8.564$, so $\mathbf{n}\approx(0.8174,\,-0.5760)$.

**6. Offset.** $c=-\mathbf{n}^T\bar{\mathbf{x}}=-(0.8174\cdot2.5-0.5760\cdot4)=-(2.0435-2.304)\approx0.2607$.

**Line:** $0.8174x-0.5760y+0.2607=0$. **Distances:** for $(1,2)$, $0.8174-1.152+0.2607\approx-0.074$, and so on. The sum of squares is $\approx0.067=\lambda_{\min}$ ✓.

# The same in Python / NumPy (2D line and 3D plane)

```python
import numpy as np

# --- 1) point-plane distance ---
l, m = np.array([2.0, 1.0, -2.0]), 3.0          # plane 2x + y - 2z + 3 = 0
x0 = np.array([3.0, 4.0, 5.0])
print("distance =", abs(l @ x0 + m) / np.linalg.norm(l))

# --- 2) optimal (geometric) line fit in 2D ---
X = np.array([[1, 2], [2, 3], [3, 5], [4, 6]], dtype=float)
centroid = X.mean(axis=0)
Xc = X - centroid                      # move centre of gravity to the origin
S = Xc.T @ Xc                          # scatter matrix
val, vec = np.linalg.eig(S)
val, vec = val.real, vec.real          # eig may return complex dtype
n = vec[:, np.argmin(val)]             # normal = smallest eigenvector
if n[0] < 0:                           # sign is arbitrary; fix it to match the paper
    n = -n
c = -n @ centroid
print("centroid =", centroid)
print("S =\n", S)
print("eigenvalues =", np.round(val, 4))
print("normal n =", np.round(n, 4), " c =", round(c, 4))
d = X @ n + c
print("signed distances =", np.round(d, 4), " sum d^2 =", round(np.sum(d**2), 4))

# --- 3) the same for a plane in 3D (synthetic noisy plane z = 0.5x - y + 2) ---
rng = np.random.default_rng(0)
xy = rng.uniform(-5, 5, size=(200, 2))
z = 0.5 * xy[:, 0] - xy[:, 1] + 2 + rng.normal(0, 0.05, 200)
P = np.column_stack([xy, z])

centroid = P.mean(axis=0)
val, vec = np.linalg.eig((P - centroid).T @ (P - centroid))
val, vec = val.real, vec.real
n = vec[:, np.argmin(val)]
n = n / np.linalg.norm(n)
d = -n @ centroid
true_n = np.array([0.5, -1, -1]) / np.linalg.norm([0.5, -1, -1])
print("fitted n =", np.round(n, 4), " d =", round(d, 4))
print("true   n =", np.round(true_n, 4), " (same up to sign)")
print("RMS distance =", round(np.sqrt(np.mean((P @ n + d)**2)), 4))
```

## Two ways to fit a plane (both are in Lagrange-1.pdf)

**Algebraic** (Section 3): rows $[x_i,\ y_i,\ z_i,\ 1]$, then the smallest eigenvector of $A^TA$ gives $(a,b,c,d)$. It is simple, but it mixes coordinate units with the "1" column, so it needs data normalization.\
**Geometric** (Section 4): subtract the centroid, then the smallest eigenvector of $X^TX$. It minimises true perpendicular distances, and this is what the solution's `refine_plane` does.

## Where this is used in the assignment

`refine_plane(P)`: centroid, then `A.T @ A`, then the smallest eigenvector. It runs after RANSAC on all inliers. The **normals** of the range image come from the same computation on the points in a small $7\times7$ window.
