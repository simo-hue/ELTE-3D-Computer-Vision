## Idea

A matrix is a linear map. $A\mathbf{x}$ is a list of dot products (row $i$ of $A$ with $\mathbf{x}$). In this course matrices do three jobs: they **transform** points (rotation $R$, camera $K$, homography $H$), they **collect equations** (one row per point, $A\mathbf{h}=\mathbf{0}$), and they **summarise data** (scatter matrix $X^TX$).

## Key formulas

## Key formulas

### 1. Matrix transpose and inverse

$$
(AB)^T = B^T A^T
$$

$$
(AB)^{-1} = B^{-1}A^{-1}
$$

**2×2 matrix inverse:**

$$
A = \begin{bmatrix}a&b\\c&d\end{bmatrix}
$$

$$
A^{-1} = \frac{1}{ad-bc}
\begin{bmatrix}d&-b\\-c&a\end{bmatrix}
\qquad (\det(A)\neq 0)
$$

### 2. Matrix rank

$$
\operatorname{rank}(A) =
\text{number of linearly independent rows (= columns)}
$$

For a square matrix:

$$
\det(A)\neq 0 \iff \operatorname{rank}(A)=n
$$

$$
\det(A)=0 \iff \operatorname{rank}(A)<n
$$

### 3. Rotation matrices

**Properties:**

$$
R^TR=I
$$

$$
\det(R)=+1
$$

$$
R^{-1}=R^T
$$

**Rotation around the Z axis:**

$$
R_z(\theta)=
\begin{bmatrix}
\cos\theta&-\sin\theta&0\\
\sin\theta&\cos\theta&0\\
0&0&1
\end{bmatrix}
$$

**Rotation around the X axis:**

$$
R_x(\theta)=
\begin{bmatrix}
1&0&0\\
0&\cos\theta&-\sin\theta\\
0&\sin\theta&\cos\theta
\end{bmatrix}
$$

**Rotation around the Y axis:**

$$
R_y(\theta)=
\begin{bmatrix}
\cos\theta&0&\sin\theta\\
0&1&0\\
-\sin\theta&0&\cos\theta
\end{bmatrix}
$$

### 4. Symmetric matrices

$$
A=A^T
$$

For any matrix \(A\):

$$
\boxed{A^TA \text{ is symmetric}}
$$

Because:

$$
(A^TA)^T=A^T(A^T)^T=A^TA
$$

# Worked example (on paper)

**(a) Inverse.** $A=\begin{bmatrix}1&2\\3&4\end{bmatrix}$:  $\det A=1\cdot4-2\cdot3=-2$, $$A^{-1}=\frac1{-2}\begin{bmatrix}4&-2\\-3&1\end{bmatrix}=\begin{bmatrix}-2&1\\1.5&-0.5\end{bmatrix},\qquad
AA^{-1}=\begin{bmatrix}1\cdot(-2)+2\cdot1.5 & 1\cdot1+2\cdot(-0.5)\\ 3\cdot(-2)+4\cdot1.5 & 3\cdot1+4\cdot(-0.5)\end{bmatrix}=I\ ✓$$

**(b) Rotation.** $\theta=90^\circ$: $\cos=0,\ \sin=1$, so $R_z=\begin{bmatrix}0&-1&0\\1&0&0\\0&0&1\end{bmatrix}$,  $R_z\begin{bmatrix}1\\0\\0\end{bmatrix}=\begin{bmatrix}0\\1\\0\end{bmatrix}$ (the $x$ axis turns into the $y$ axis). Columns are unit length and perpendicular, so $R_z^TR_z=I$; $\det R_z=0\cdot0-(-1)\cdot1=1$.

**(c) Rank.** $M=\begin{bmatrix}1&2&3\\2&4&6\\1&0&1\end{bmatrix}$: row 2 $=2\times$ row 1, so at most 2 independent rows. Rows 1 and 3 are not multiples $\Rightarrow$ $\text{rank}(M)=2$, $\det M=0$.

**(d) Many dot products at once.** Stack points as rows, $P\in\mathbb{R}^{N\times3}$. Then $P\mathbf{n}$ is the vector of all $\mathbf{p}_i^T\mathbf{n}$. With the plane from topic 01 ($\mathbf{n}=(6,3,2)/7,\ d=-6/7$): for $(1,0,0)$: $6/7-6/7=0$; for $(0,2,0)$: $6/7-6/7=0$; for $(1,1,1)$: $11/7-6/7=5/7\approx0.714$.

# The same in Python / NumPy

```python
import numpy as np

# inverse of a 2x2 matrix
A = np.array([[1.0, 2.0],
              [3.0, 4.0]])
print("det(A) =", round(np.linalg.det(A), 6))
print("inv(A) =\n", np.linalg.inv(A))
print("A @ inv(A) =\n", np.round(A @ np.linalg.inv(A), 10))

# rotation about the z axis by 90 degrees
t = np.radians(90)
Rz = np.array([[np.cos(t), -np.sin(t), 0],
               [np.sin(t),  np.cos(t), 0],
               [0,          0,         1]])
print("Rz @ [1,0,0] =", np.round(Rz @ np.array([1, 0, 0]), 10))
print("Rz.T @ Rz =\n", np.round(Rz.T @ Rz, 10))
print("det(Rz) =", round(np.linalg.det(Rz), 10))

# rank: the 2nd row is 2 * the 1st row
M = np.array([[1, 2, 3],
              [2, 4, 6],
              [1, 0, 1]])
print("rank(M) =", np.linalg.matrix_rank(M))

# many dot products at once: (N x 3) @ (3,) -> (N,)
P = np.array([[1, 0, 0],
              [0, 2, 0],
              [1, 1, 1]], dtype=float)
n = np.array([6, 3, 2]) / 7.0
d = -6 / 7.0
print("P @ n + d =", np.round(P @ n + d, 4))
```

### Exercise 3

Is $R_z(30^\circ)R_x(90^\circ)=R_x(90^\circ)R_z(30^\circ)$? Check with `np.allclose`.

> **Expected result:** `False`. $R_zR_x=\begin{bmatrix}0.866&0&0.5\\0.5&0&-0.866\\0&1&0\end{bmatrix}$ and $R_xR_z=\begin{bmatrix}0.866&-0.5&0\\0&0&-1\\0.5&0.866&0\end{bmatrix}$.

```python
# Exercise 3: your solution
t_z = np.radians(30)
t_x = np.radians(90)

Rz = np.array([
    [np.cos(t_z), -np.sin(t_z), 0],
    [np.sin(t_z),  np.cos(t_z), 0],
    [0, 0, 1]
])

Rx = np.array([
    [1, 0, 0],
    [0, np.cos(t_x), -np.sin(t_x)],
    [0, np.sin(t_x),  np.cos(t_x)]
])

result1 = Rz @ Rx
result2 = Rx @ Rz

print(np.allclose(result1, result2))

# np.allclose(a, b) checks whether two arrays are approximately equal, allowing for small floating-point differences.
a = np.array([0.1 + 0.2])
b = np.array([0.3])

print(a == b)          # [False]
print(np.allclose(a, b))  # True
```
