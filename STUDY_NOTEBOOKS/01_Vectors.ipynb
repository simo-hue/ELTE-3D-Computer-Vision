## Idea

A point in 3D is a vector $\mathbf{p}=[x,y,z]^T$. Almost all of the geometry in this course comes down to three operations: the **dot product** (angles, projections, distances), the **norm** (length), and the **cross product** (a vector perpendicular to two others, e.g. the normal of a plane). A plane is described by a unit normal $\mathbf{n}$ and an offset $d$: $\mathbf{n}^T\mathbf{p}+d=0$.

## Key formulas

### 1. Dot product

$$
\mathbf{a}\cdot\mathbf{b}
= \mathbf{a}^T\mathbf{b}
= a_1b_1+a_2b_2+a_3b_3
= \|\mathbf{a}\|\|\mathbf{b}\|\cos\theta
$$

### 2. Norm and unit vector

$$
\|\mathbf{a}\|=\sqrt{\mathbf{a}^T\mathbf{a}}
$$

$$
\hat{\mathbf{a}}=\frac{\mathbf{a}}{\|\mathbf{a}\|}
\qquad \text{(unit vector)}
$$

### 3. Cross product

$$
\mathbf{a}\times\mathbf{b}
=
\begin{bmatrix}
a_2b_3-a_3b_2\\
a_3b_1-a_1b_3\\
a_1b_2-a_2b_1
\end{bmatrix}
$$

The result is perpendicular to both input vectors:

$$
\mathbf{a}\times\mathbf{b}\perp\mathbf{a},
\qquad
\mathbf{a}\times\mathbf{b}\perp\mathbf{b}
$$

Its length is:

$$
\|\mathbf{a}\times\mathbf{b}\|
=
\|\mathbf{a}\|\|\mathbf{b}\|\sin\theta
$$

### 4. Plane through three points

$$
\mathbf{n}=
\frac{
(\mathbf{p}_2-\mathbf{p}_1)\times
(\mathbf{p}_3-\mathbf{p}_1)
}{
\|(\mathbf{p}_2-\mathbf{p}_1)\times
(\mathbf{p}_3-\mathbf{p}_1)\|
}
$$

$$
d=-\mathbf{n}^T\mathbf{p}_1
$$

Plane equation:

$$
\boxed{\mathbf{n}^T\mathbf{p}+d=0}
$$

### 5. Signed distance

For a point \(\mathbf{q}\):

$$
\boxed{\text{Signed distance}=\mathbf{n}^T\mathbf{q}+d}
$$

Only when \(\|\mathbf{n}\|=1\).

Absolute distance:

$$
\boxed{\text{Distance}=|\mathbf{n}^T\mathbf{q}+d|}
$$

### Important

If the three points are on one line (collinear), their cross product is the zero vector:

$$
\mathbf{n}=\mathbf{0}
$$

No unique plane can be defined (degenerate sample).


# Worked example (on paper)

Plane through $A=(1,0,0)$, $B=(0,2,0)$, $C=(0,0,3)$.

**1. Edge vectors.**  $\mathbf{AB}=B-A=(-1,2,0)$,  $\mathbf{AC}=C-A=(-1,0,3)$.

**2. Angle between them.** $$\mathbf{AB}\cdot\mathbf{AC}=(-1)(-1)+2\cdot0+0\cdot3=1,\quad \|\mathbf{AB}\|=\sqrt5,\ \|\mathbf{AC}\|=\sqrt{10}$$ $$\cos\theta=\frac{1}{\sqrt5\sqrt{10}}=\frac1{\sqrt{50}}\approx0.1414\ \Rightarrow\ \theta\approx81.87^\circ$$

**3. Normal = cross product.** $$\mathbf{n}=\mathbf{AB}\times\mathbf{AC}=
\begin{bmatrix} 2\cdot3-0\cdot0\\ 0\cdot(-1)-(-1)\cdot3\\ (-1)\cdot0-2\cdot(-1)\end{bmatrix}
=\begin{bmatrix}6\\3\\2\end{bmatrix},\qquad \|\mathbf{n}\|=\sqrt{36+9+4}=7$$

**4. Unit normal and offset.**  $\hat{\mathbf{n}}=\left(\tfrac67,\tfrac37,\tfrac27\right)$,  $d=-\hat{\mathbf{n}}^TA=-\tfrac67$.   Plane: $\tfrac67x+\tfrac37y+\tfrac27z-\tfrac67=0$  (or $6x+3y+2z=6$).

**5. Check.** $B$: $\tfrac37\cdot2-\tfrac67=0$ ✓,  $C$: $\tfrac27\cdot3-\tfrac67=0$ ✓.

**6. Distance of the origin.** $|\hat{\mathbf{n}}^T\mathbf{0}+d|=\tfrac67\approx0.857$.

# The same in Python / NumPy

```python
import numpy as np

# three points of a plane
A = np.array([1.0, 0.0, 0.0])
B = np.array([0.0, 2.0, 0.0])
C = np.array([0.0, 0.0, 3.0])

AB = B - A
AC = C - A
print("AB =", AB, " AC =", AC)

# dot product, norms, angle
dot = np.dot(AB, AC)
cos_angle = dot / (np.linalg.norm(AB) * np.linalg.norm(AC))
print("AB . AC =", dot)
print("angle   =", round(np.degrees(np.arccos(cos_angle)), 2), "deg")

# normal of the plane = cross product
n = np.cross(AB, AC)
print("n = AB x AC =", n, " |n| =", np.linalg.norm(n))

n_unit = n / np.linalg.norm(n)
d = -np.dot(n_unit, A)
print("unit n =", np.round(n_unit, 4), " d =", round(d, 4))

# every point of the plane gives n.p + d = 0
for name, p in [("A", A), ("B", B), ("C", C)]:
    print(name, "->", round(np.dot(n_unit, p) + d, 10))

# signed distance of the origin (|n| = 1, so n.p + d IS the distance)
print("distance of origin =", round(abs(np.dot(n_unit, [0, 0, 0]) + d), 4))
```

### Exercise 5

Why do $(0,0,0)$, $(1,1,1)$, $(2,2,2)$ not define a plane? What does `np.cross` return for them?

```python
# Exercise 5: your solution
a = np.array([0.0, 0.0, 0.0])
b = np.array([1.0, 1.0, 1.0])
c = np.array([2.0, 2.0, 2.0])

ab = b - a
ac = c - a

cross = np.cross(ab, ac)
print(cross)

cos = np.dot(ab, ac) / (
    np.linalg.norm(ab) * np.linalg.norm(ac)
)

cos = np.clip(cos, -1.0, 1.0)
print(np.degrees(np.arccos(cos)))
```
